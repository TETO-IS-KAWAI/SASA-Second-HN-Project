```python
import PIL
from PIL import Image
import os
```

```python
os.makedirs("img/img-resized", exist_ok=True)
```

```python
for i in range(1, 6):
    image=Image.open(f"img/origin/img-{i}.jpg")
    image_resized=image.resize((3200, 3200))
    image_resized.save(f"img/img-resized/img-{i}-resized.jpg")
    os.makedirs(f"img/img-{i}-resized", exist_ok=True)
    for j in range(0, 4):
        for k in range(0, 4):
            image_resized.crop((k*800, j*800, (k+1)*800, (j+1)*800)).save(f"img/img-{i}-resized/img-{i}-resized-{j*4+k+1}.jpg")
```
