```python
import cv2
import numpy
import pandas
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from skimage import io
from PIL import Image
import os
import matplotlib.patches as mpatch
```

```python
os.makedirs("append-cluster", exist_ok=True)
os.makedirs("append-cluster/image_after")
os.makedirs("append-cluster/image_after/5")
```

```python
os.makedirs("append-cluster/data_after/5", exist_ok=True)
```

```python
def Kmeans_color_segment(image_path, n, x, z):
    image=io.imread(image_path)
    # image_after=cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image_after=image
    
    pixel_data = image_after.reshape((-1, 3))
    pixel_data = numpy.float32(pixel_data)
    
    kmeans = KMeans(n_clusters=n, random_state=None, init='k-means++', n_init='auto')
    kmeans.fit(pixel_data)
    
    labels = kmeans.labels_
    centers = numpy.uint8(kmeans.cluster_centers_)
    segmented_image = centers[labels].reshape(image_after.shape)
    cluster_areas = {i: numpy.sum(labels == i) for i in range(n)}
    
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(image_after, cv2.COLOR_BGR2RGB))
    plt.title('Original Image')
    plt.legend()
    
    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(segmented_image, cv2.COLOR_BGR2RGB))
    plt.title('Segmented Image (KMeans)')
    plt.savefig(f"append-cluster/image_after/5/img-{x}-retouched-{z}.jpg")
    plt.close()
    
    print("각 클러스터의 면적(픽셀 수):", cluster_areas)
    print("클러스터 중심 색상 (RGB):", centers)
    
    data = {'cluster_id': [], 'rgb_color': [], 'area_pixels': []}
    
    for i in range(n):
        data['cluster_id'].append(f'Cluster {i}')
        data['rgb_color'].append(tuple(centers[i]))
        data['area_pixels'].append(cluster_areas[i])
        
    df = pandas.DataFrame(data)
    df.to_csv(f"append-cluster/data_after/5/retouched-data-{x}-{z}.csv", index=False)
    print(f"클러스터 면적 정보가 저장되었습니다.")
    
    return cluster_areas, segmented_image
```

```python
for i in range(1, 6):
    Kmeans_color_segment(f"img/img-resized/img-{i}-resized.jpg", 5, i, 5)
```
